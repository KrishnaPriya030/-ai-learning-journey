```python
PASS_MARK = 50
students={
    "Krishna":100,
    "Ram":95,
    "Radha":15,
    "Ramesh":95,
    "George":100,
    "Arjun":100,
    "den":30,
    "Riya":29,
    "Shyam":15
}
passed=[student for student,mark in  students.items() if mark > PASS_MARK ]
failed=[ student for student,mark in  students.items() if mark < PASS_MARK ]


print("*" * 50)
print("Student marks details")
print("*" * 50)
print(f"Total number of students: {len(students)}")
print(f"Total number of passed students: {len(passed)}")
print(f"Total number of failed students: {len(failed)}")
print("*" * 50)
print("*" * 50)
print(f"Passed:{passed}")
print(f"Failed:{failed}")
print("*" * 50)
print("*" * 50)

for student,mark in students.items():

   status="Pass" if mark >= PASS_MARK else "Fail"
   print(f"{student} -> {mark} -> {status}")
print("*" * 50)
print("*" * 50)


```

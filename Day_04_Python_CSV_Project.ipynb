```python
import random
import csv
import math

random.seed(42)
students=["Ram","Raj","Sheen","Sai","Shyam","Savioe","Seeta","Deen"]
marks=[]
with open("myfile.csv","w",newline="") as file:
  fields=["Name","Mark","Grade","Status"]
  writer=csv.DictWriter(file,fieldnames=fields)
  writer.writeheader()

  for student in students:
     mark=random.randint(20,50)
     marks.append(mark)
     grade="A" if mark>30 else "B" if mark>25 else "C"
     status="Pass" if mark>30 else "Fail"


     writer.writerow({
     "Name" : student,
      "Mark" : mark,
     "Grade" : grade,
     "Status" : status
      })

average=sum(marks)/len(marks)
highest=max(marks)
lowest=min(marks)
sqrt_avg=math.sqrt(average)
round_high=math.ceil(highest)
round_low=math.floor(lowest)



try:
  with open("myfile.csv","r") as file:
     reader=csv.DictReader(file)


     for row in reader:
       print(f"{row['Name']:10} -> "
       f"{row['Mark']:5} -> "
       f"{row['Grade']:5} -> "
       f"{row['Status']:5}")


  print(f"Average->{sqrt_avg}")
  print(f"highest->{round_high}")
  print(f"lowest->{round_low}")


except FileNotFoundError:
  print("File not exist")
```

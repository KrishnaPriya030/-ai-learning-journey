```python
import math
import csv

class Student:
  def __init__(self,name,marks,subject):
    self.name=name
    self.marks=marks
    self.subject=subject

  def average(self):
    try:
      average_marks=sum(self.marks)/len(self. marks)
      return round(average_marks, 2)
    except ZeroDivisionError:
        return 0
  def grade(self):
    avg = self.average()
    if avg >= 90:
        return "A"
    elif avg >= 50:
        return "B"
    else:
        return "C"




class AIStudent(Student):
  def __init__(self,name,marks,subject,project_name):
    super().__init__(name,marks,subject)
    self.project_name=project_name
  def analyze_performance(self):
    print(f"Name:{self.name}")
    print(f"Marks:{self.marks}")
    print(f"subject:{self.subject}")
    print(f"average:{self.average()}")
    print(f"project:{self.project_name}")
    print(f"grade:{self.grade()}")



s1=AIStudent("Krishna", [85,92,78],
               "AI", "Chatbot")
s2=AIStudent("Ram", [45,38,52],
               "ML", "Detector")
s3=AIStudent("Priya", [92,95,88],
               "DL", "ImageNet")

all_students=[s1,s2,s3]

try:
  with open("students.csv", "w", newline="") as file:
    fields = ["Name", "Marks", "Subject",
              "Average", "Grade", "Project"]
    writer = csv.DictWriter(file, fieldnames=fields)
    writer.writeheader()

    for s in all_students:
      writer.writerow({
          "Name": s.name,
          "Marks":s.marks,
          "Subject":s.subject,
          "Average":s.average(),
          "Grade":s.grade(),
          "Project":s.project_name
      })

except PermissionError:
  print("❌ Cannot write file!")
except Exception as e:
    print(f"❌ Error saving: {e}")

print("="*50)
print("   📊 AI STUDENT PERFORMANCE SYSTEM")
print("="*50)


try:
  with open("students.csv","r") as file:
    reader=csv.DictReader(file)

    for row in reader:
          print(f"{row['Name']:10}->"
          f"{row['Average']:10}->"
          f"{row['Grade']:5}->"
          f"{row['Project']:10}"

    )
except FileNotFoundError:
  print("❌ File not found!")
except Exception as e:
    print(f"❌ Error reading: {e}")


print("="*50)
print("🏆 RANKINGS:")
print("="*50)
ranked=sorted(all_students,key=lambda x:x.average(),reverse=True)

for i,model in enumerate(ranked,1):
  print(f"{i}.{model.name}->{model.average()}->{model.project_name}")





passed=[s.name for s in all_students if s.average()>=50]
failed=[s.name for s in all_students if s.average()<50]

```

```python
students = {

      "Hari" : 25,
      "sachin" : 50,
      "Arjun" : 50,
      "Hima" : 45,
      "Shreya" : 25,
      "Deepu" : 20,
      "laya" : 46
}

def evaluate(data,pass_mark=30,rankings=True):

  try:
    if data=={}:
      return "Empty dictionary!"
    failed =[name for name,mark in students.items() if mark< pass_mark]


    return len(data.values()),sum(data.values())/len(data.values()),max(data.values()),min(data.values()),failed

  except TypeError:
    return "❌ Invalid data format!"




total_students,average_marks,maximum_marks,minimum_marks,fail = evaluate(students)

rank = sorted(students.items(),key=lambda x:x[1],reverse=True)
print("="*50)

print("🏆 Rankings:")

print("="*50)

for i,(name,mark) in enumerate(rank,1):
  print(f"{i}.{name} -> {mark}")

grade = dict(map(lambda item:(item[0], "A" if item[1] > 45 else "B" if item[1] >25 else "C"),students.items()))

print("="*50)
print("   📊 STUDENT ANALYZER")
print("="*50)
print(f"{grade}")
print(f"{fail}")
print(f"{total_students}")
print(f"{average_marks}")
print(f"{maximum_marks}")
print(f"{minimum_marks }")
print("="*50)
print("="*50)


```

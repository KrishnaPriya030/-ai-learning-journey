```python
# ================================================
# 🤖 AI Learning Journey - Day 1
# 📅 Date:09.06.2026
# 📌 Topic: Python Fundamentals
# ================================================

# Student Grade Classifier
# Uses: Variables, Lists, Functions,
#       Loops, Conditions

students = ["Nima", "Hima", "Sharukh"]
marks = [20, 50, 60]

def get_grade(mark):
    if mark > 55:
        return "A"
    elif mark > 30:
        return "B"
    else:
        return "Failed"

for i, student in enumerate(students):
    grade = get_grade(marks[i])
    print(f"{student} -> {marks[i]} -> Grade {grade}")
```

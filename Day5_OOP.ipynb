```python
class AIModel:
  def __init__(self,name,accuracy,epochs):
    self.name=name
    self.accuracy=accuracy
    self.epochs=epochs
  def train(self):
    print(f"{self.name} is trained well")
  def evaluate(self):
    print(f"The model's {self.accuracy} is based on their training")
  def display(self):
    print(f"{self.name} -> {self.accuracy} -> {self.epochs}")

class CNNModel(AIModel):
  def __init__(self,name,accuracy,epochs,filters):
    super().__init__(name,accuracy,epochs)
    self.filters=filters
  def evaluate(self):
    print(f"The model's {self.accuracy} is based on their training and the cnn model is famous one!")
  def detect_objects(self):
    print(f"The model can detect objects easily")

class NLPModel(AIModel):
   def __init__(self,name,accuracy,epochs,vocab_size):
    super().__init__(name,accuracy,epochs)
    self.vocab_size=vocab_size
   def evaluate(self):
    print(f"The model's {self.accuracy} is based on their training and {self.vocab_size} the cnn model is famous one!")

   def process_text(self):
    print(f"The method processess text!")

ai1=AIModel("Ai",100,2.5)
ai2=AIModel("Ai",50,2.5)
ai1.display()
ai2.display()

cn1=CNNModel("cnmodel",100,5,50)
cn2=CNNModel("cnmodel2",50,5,25)
cn1.display()
cn2.display()

nlp1=NLPModel("nlpmodel1",4,82,6)
nlp=NLPModel("nlpmodel2",8,69,3)
nlp1.display()
nlp.display()

all_models = [ai1, ai2, cn1, cn2, nlp1, nlp]
rank=sorted(all_models,key=lambda x:x.accuracy,reverse=True)
for i,model in enumerate(rank,1):
  print(f"{i}. {model.name:10} → {model.accuracy}")






```
